```python

import pandas as pd

CHEXPERT_ROOT = "../data/raw/chexpert"

train_df = pd.read_csv(f"{CHEXPERT_ROOT}/train.csv")
valid_df = pd.read_csv(f"{CHEXPERT_ROOT}/valid.csv")

print(train_df.shape)
print(valid_df.shape)
```

```python
print(train_df["Frontal/Lateral"].value_counts())
print(train_df["AP/PA"].value_counts())

pathology_cols = [c for c in train_df.columns 
                   if c not in ["Path", "Sex", "Age", "Frontal/Lateral", "AP/PA"]]
print(pathology_cols)

for col in pathology_cols:
    print(col, train_df[col].value_counts(dropna=False).to_dict())
```
