```python
import pandas as pd

ISIC_ROOT = "../data/raw/isic"

isic_df = pd.read_csv(f"{ISIC_ROOT}/ISIC_2019_Training_GroundTruth.csv")  
print(isic_df.shape)
isic_df.head()
```

```python
class_cols = [c for c in isic_df.columns if c != "image"]
print(class_cols)

class_counts = isic_df[class_cols].sum().sort_values(ascending=False)
print(class_counts)
```

```python
isic_df["label"] = isic_df[class_cols].idxmax(axis=1)
print(isic_df["label"].value_counts())
```
